```python
import numpy as np
import matplotlib.pyplot as plt


x = [0, 1, 2]
y = [1.5, 1, 2]

fig, ax = plt.subplots(figsize=(8, 5))

# Plot data points
ax.scatter(x, y, s=80, zorder=5)
ax.text(x[0] - 0.75, y[0] + 0.35, f"({x[0]}, {y[0]})", fontsize=9)
ax.text(x[1] + 0.15, y[1] - 0.35, f"({x[1]}, {y[1]})", fontsize=9)
ax.text(x[2] + 0.15, y[2] - 0.35, f"({x[2]}, {y[2]})", fontsize=9)

ax.axhline(y=0, color='k', linewidth=0.5)
ax.axvline(x=0, color='k', linewidth=0.5)

ax.axvline(x=-1, color='red', linewidth=1)
ax.text(-0.8, -1.5, "y(-1) = ?", color='red', fontsize=9)




ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_xlim(-2, 3)
ax.set_ylim(-2, 3)
ax.set_aspect('equal')
ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.6)
ax.set_xticks(np.arange(-2, 4, 0.5))
ax.set_yticks(np.arange(-2, 4, 0.5))

plt.title("Data points")
plt.show()
```

```python
def l2_error(m, c):
    error = 0
    for i in range(len(x)):
        error += (y[i] - (m * x[i] + c)) ** 2
    return error

def l1_error(m, c):
    error = 0
    for i in range(len(x)):
        error += abs(y[i] - (m * x[i] + c))
    return error
```

```python
# plot contours of L2 error
m_values = np.linspace(-3, 5, 500)
c_values = np.linspace(-3, 5, 500)
M, C = np.meshgrid(m_values, c_values)
L2_errors = np.array([[l2_error(m, c) for m in m_values] for c in c_values])

fig, ax = plt.subplots(figsize=(8, 6))

# Create contour plot
contour_lines = ax.contour(M, C, L2_errors, levels=15, colors='black', linewidths=0.7)
ax.clabel(contour_lines, inline=True, fontsize=8)

# Find and mark the minimum
min_idx = np.unravel_index(np.argmin(L2_errors), L2_errors.shape)
min_m = m_values[min_idx[1]]
min_c = c_values[min_idx[0]]
min_error = L2_errors[min_idx]

ax.plot(min_m, min_c, 'r*', markersize=20, label=f'Minimum: m={min_m:.3f}, c={min_c:.3f}')
ax.legend()

ax.set_xlabel('m (slope)')
ax.set_ylabel('c (intercept)')
ax.set_title('L2 Error Contour Plot')
plt.show()
```

```python
# plot contours of L1 error
m_values = np.linspace(-3, 5, 500)
c_values = np.linspace(-3, 5, 500)
M, C = np.meshgrid(m_values, c_values)
L1_errors = np.array([[l1_error(m, c) for m in m_values] for c in c_values])

fig, ax = plt.subplots(figsize=(8, 6))

# Create contour plot
contour_lines = ax.contour(M, C, L1_errors, levels=15, colors='black', linewidths=0.7)
ax.clabel(contour_lines, inline=True, fontsize=8)

# Find and mark the minimum
min_idx = np.unravel_index(np.argmin(L1_errors), L1_errors.shape)
min_m = m_values[min_idx[1]]
min_c = c_values[min_idx[0]]
min_error = L1_errors[min_idx]

ax.plot(min_m, min_c, 'r*', markersize=20, label=f'Minimum: m={min_m:.3f}, c={min_c:.3f}')
ax.legend()

ax.set_xlabel('m (slope)')
ax.set_ylabel('c (intercept)')
ax.set_title('L1 Error Contour Plot')
plt.show()
```

```python
# Find optimal parameters for both L2 and L1
m_values = np.linspace(-3, 5, 500)
c_values = np.linspace(-3, 5, 500)

# L2 minimum
L2_errors = np.array([[l2_error(m, c) for m in m_values] for c in c_values])
min_idx_L2 = np.unravel_index(np.argmin(L2_errors), L2_errors.shape)
m_L2 = m_values[min_idx_L2[1]]
c_L2 = c_values[min_idx_L2[0]]

# L1 minimum
L1_errors = np.array([[l1_error(m, c) for m in m_values] for c in c_values])
min_idx_L1 = np.unravel_index(np.argmin(L1_errors), L1_errors.shape)
m_L1 = m_values[min_idx_L1[1]]
c_L1 = c_values[min_idx_L1[0]]

# Plot scatter and regression lines
fig, ax = plt.subplots(figsize=(8, 6))

# Scatter plot of original data
ax.scatter(x, y, s=100, color='black', zorder=5, label='Data points')

# L2 regression line
x_line = np.linspace(-1, 3, 100)
y_L2 = m_L2 * x_line + c_L2
ax.plot(x_line, y_L2, 'b-', linewidth=2, label=f'L2: y={m_L2:.3f}x+{c_L2:.3f}')

# L1 regression line
y_L1 = m_L1 * x_line + c_L1
ax.plot(x_line, y_L1, 'r-', linewidth=2, label=f'L1: y={m_L1:.3f}x+{c_L1:.3f}')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Data Points with L2 and L1 Regression Lines')
ax.legend()
ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.6)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)


n = 100

# Group 1: low x, high y
x1 = np.random.normal(2, 1, n)
y1 = 1.5 * x1 + 12 + np.random.normal(0,2, n)

# Group 2: medium x, medium y  
x2 = np.random.normal(4.5, 1, n)
y2 = 1.5 * x2 + 4 + np.random.normal(0, 2, n)

# Group 3: high x, low y
x3 = np.random.normal(7, 1, n)
y3 = 1.5 * x3 - 6 + np.random.normal(0,2, n)

# Combined data
x_all = np.concatenate([x1, x2, x3])
y_all = np.concatenate([y1, y2, y3])

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x1, y1, alpha=0.6, s=40, color='k')
ax.scatter(x2, y2, alpha=0.6, s=40, color='k')
ax.scatter(x3, y3, alpha=0.6, s=40, color='k')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True, linestyle='--', alpha=0.5)

y_min = min(min(y1), min(y2), min(y3)) - 1
y_max = max(max(y1), max(y2), max(y3)) + 1
ax.set_ylim(y_min, y_max)
ax.set_xlim(0, 10)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Combined data
x_all = np.concatenate([x1, x2, x3])
y_all = np.concatenate([y1, y2, y3])

# Plot 1: Colored by group with group regression lines
ax = axes[0]
colors = ['#e41a1c', '#377eb8', '#4daf4a']
labels = ['Group A', 'Group B', 'Group C']

for x_g, y_g, color, label in [(x1, y1, colors[0], labels[0]), 
                                (x2, y2, colors[1], labels[1]), 
                                (x3, y3, colors[2], labels[2])]:
    ax.scatter(x_g, y_g, c=color, alpha=0.6, s=40, label=label)
    m, c = np.polyfit(x_g, y_g, 1)
    x_line = np.linspace(min(x_g) - 0.3, max(x_g) + 0.3, 50)
    ax.plot(x_line, m * x_line + c, color=color, linewidth=2)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Within Each Group: Positive Slope')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)

# Plot 2: Combined with overall regression line
ax = axes[1]
ax.scatter(x_all, y_all, c='gray', alpha=0.6, s=40)

m_all, c_all = np.polyfit(x_all, y_all, 1)
x_line = np.linspace(0, 10, 100)
ax.plot(x_line, m_all * x_line + c_all, 'k-', linewidth=2.5, label=f'Overall slope = {m_all:.2f}')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Combined Data: Negative Slope!')
ax.legend()
ax.grid(True, linestyle='--', alpha=0.5)

y_min = min(min(y1), min(y2), min(y3)) - 1
y_max = max(max(y1), max(y2), max(y3)) + 1
for ax in axes:
    ax.set_ylim(y_min, y_max)
    ax.set_xlim(0, 10)

plt.tight_layout()
plt.show()
```

```python
from ipywidgets import interactive, FloatSlider
import numpy as np

import matplotlib.pyplot as plt

# Ten points that mostly follow a line
x_data = np.arange(10, dtype=float)
base_y_data = 0.8 * x_data + 1 + np.random.normal(0, 1, 10)
outlier_indices = (2, 8)
y_data = base_y_data.copy()

def current_l2_error(x_values, y_values, m, c):
  return np.sum((y_values - (m * x_values + c)) ** 2)

def current_l1_error(x_values, y_values, m, c):
  return np.sum(np.abs(y_values - (m * x_values + c)))

def fit_l1_line(x_values, y_values):
  candidate_slopes = []
  for i in range(len(x_values)):
    for j in range(i + 1, len(x_values)):
      if x_values[i] != x_values[j]:
        candidate_slopes.append((y_values[j] - y_values[i]) / (x_values[j] - x_values[i]))

  if not candidate_slopes:
    return 0.0, float(np.median(y_values))

  candidate_slopes = np.array(sorted(set(candidate_slopes)))
  test_slopes = candidate_slopes.tolist()
  test_slopes.append(candidate_slopes[0] - 1)
  test_slopes.append(candidate_slopes[-1] + 1)
  test_slopes.extend(((candidate_slopes[:-1] + candidate_slopes[1:]) / 2).tolist())

  best_m = None
  best_c = None
  best_error = None

  for slope in test_slopes:
    intercept = float(np.median(y_values - slope * x_values))
    error = current_l1_error(x_values, y_values, slope, intercept)
    if best_error is None or error < best_error:
      best_m = float(slope)
      best_c = intercept
      best_error = float(error)

  return best_m, best_c

def plot_regressions(y_outlier_1=base_y_data[outlier_indices[0]], y_outlier_2=base_y_data[outlier_indices[1]]):
  """Plot L1 and L2 regressions while moving two potential outliers."""
  
  x_current = x_data.copy()
  y_current = base_y_data.copy()
  y_current[outlier_indices[0]] = y_outlier_1
  y_current[outlier_indices[1]] = y_outlier_2
  
  # Calculate L2 fit
  m_l2, c_l2 = np.polyfit(x_current, y_current, 1)
  l2_err = current_l2_error(x_current, y_current, m_l2, c_l2)
  
  # Calculate the L1 fit for the current points
  m_l1, c_l1 = fit_l1_line(x_current, y_current)
  l1_err = current_l1_error(x_current, y_current, m_l1, c_l1)
  
  # Create figure
  fig, ax = plt.subplots(figsize=(10, 6))
  
  # Plot data points
  mask = np.ones(len(x_current), dtype=bool)
  mask[list(outlier_indices)] = False
  ax.scatter(x_current[mask], y_current[mask], s=90, color='black', zorder=5, label='Points near the line')
  ax.scatter(x_current[list(outlier_indices)], y_current[list(outlier_indices)], s=120, color='darkorange', edgecolor='black', zorder=6, label='Adjustable points')
  
  # Add point labels
  for i, (xi, yi) in enumerate(zip(x_current, y_current)):
    label_color = 'darkorange' if i in outlier_indices else 'black'
    ax.text(xi + 0.10, yi + 0.18, f'P{i}', fontsize=9, color=label_color)
  
  # Plot regression lines
  x_line = np.linspace(x_current.min() - 0.5, x_current.max() + 0.5, 200)
  y_l2 = m_l2 * x_line + c_l2
  y_l1 = m_l1 * x_line + c_l1
  
  ax.plot(x_line, y_l2, 'b-', linewidth=2.5, label=f'L2: y={m_l2:.3f}x+{c_l2:.3f}')
  ax.plot(x_line, y_l1, 'r--', linewidth=2.5, label=f'L1: y={m_l1:.3f}x+{c_l1:.3f}')
  
  # Plot error lines
  for xi, yi in zip(x_current, y_current):
    y_l2_pred = m_l2 * xi + c_l2
    y_l1_pred = m_l1 * xi + c_l1
    ax.plot([xi, xi], [yi, y_l2_pred], 'b-', alpha=0.3, linewidth=1)
    ax.plot([xi, xi], [yi, y_l1_pred], 'r--', alpha=0.3, linewidth=1)
  
  ax.set_xlabel('x', fontsize=12)
  ax.set_ylabel('y', fontsize=12)
  ax.set_title(f'L2 Error: {l2_err:.3f} | L1 Error: {l1_err:.3f}', fontsize=14, fontweight='bold')
  ax.legend(fontsize=11, loc='upper left')
  ax.grid(True, linestyle='--', alpha=0.4)
  ax.set_xlim(x_current.min() - 0.5, x_current.max() + 0.5)
  ax.set_ylim(y_current.min() - 1.0, y_current.max() + 1.0)
  
  plt.tight_layout()
  plt.show()

# Create interactive widget
interactive_plot = interactive(
  plot_regressions,
  y_outlier_1=FloatSlider(min=-10, max=20, step=0.1, value=base_y_data[outlier_indices[0]], description='P2 y'),
  y_outlier_2=FloatSlider(min=-10, max=20, step=0.1, value=base_y_data[outlier_indices[1]], description='P8 y'),
)

interactive_plot
```
